```python
from collections import defaultdict

LAST_NAME = "Fabia"
SEED_NUM = 6
FAVORITE_ARTIST = "Luke Chiang"

customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]

products = [
    "NOTEBOOK",
    "PEN",
    "USB",
    "HEADPHONES",
    "CALCULATOR"
]
```

```python
prices = {
    "NOTEBOOK": 50,
    "PEN": 15,
    "USB": 350,
    "HEADPHONES": 800,
    "CALCULATOR": 450
}
```

```python
transactions = []

artist_positions = [i + 1 for i in range(len(FAVORITE_ARTIST))]

for i, customer in enumerate(customers):

    for j, product in enumerate(products):

        quantity = (SEED_NUM +
                    artist_positions[(i + j)
                    % len(artist_positions)])

        value = quantity * prices[product]

        transactions.append({
            "Customer": customer,
            "Product": product,
            "Quantity": quantity,
            "Price": prices[product],
            "Value": value
        })

# repeated purchase
transactions.append(transactions[0].copy())
```

```python
print("GENERATED TRANSACTION RECORDS")

for transaction in transactions:
    print(transaction)
```

```python
from collections import defaultdict

counter = defaultdict(int)

for transaction in transactions:

    key = (
        transaction["Customer"],
        transaction["Product"]
    )

    counter[key] += 1

print("REPEATED PURCHASES")

for key, count in counter.items():
    if count > 1:
        print(key)
```

```python
customer_summary = defaultdict(float)

for transaction in transactions:

    customer_summary[
        transaction["Customer"]
    ] += transaction["Value"]

print("CUSTOMER SUMMARIES")

for customer, total in customer_summary.items():
    print(customer, total)
```

```python
product_summary = defaultdict(float)

for transaction in transactions:
    product_summary[
        transaction["Product"]
    ] += transaction["Value"]

print("PRODUCT SUMMARIES")

for product, total in product_summary.items():
    print(product, total)
```

```python
ordered = sorted(
    transactions,
    key=lambda x: x["Value"],
    reverse=True
)

for transaction in ordered:
    print(transaction)
```

```python
overall = sum(
    transaction["Value"]
    for transaction in transactions
)

print("FINAL CONSOLIDATED TRANSACTION REPORT")
print("Total Transactions:", len(transactions))
print("Overall Dataset Value:", overall)
```
