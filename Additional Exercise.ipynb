```python
from collections import defaultdict

LAST_NAME = "Fabia"
SEED_NUM = 6
FAVORITE_ARTIST = "Luke Chiang"

students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]

Categories = [
    "TEMPERATURE",
    "VOLTAGE",
    "CURRENT",
    "PRESSURE",
    "SPEED",
]
```

```python
artist_positions = [i + 1 for i in range(len(FAVORITE_ARTIST))]

records = []

for i, student in enumerate(students):
    for j, Categories in enumerate(Categories):

        score = ((SEED_NUM *
                 artist_positions[(i + j) % len(artist_positions)])
                 % 51) + 50

        records.append({
            "Student": student,
            "Categories": Categories,
            "Score": score
        })

# Add one repeated record
records.append(records[0].copy())

# Add one incomplete record
records.append({
    "Student": students[2],
    "Categories": "PRESSURE",
    "Score": None
})

print("Generated Records:")
for record in records:
    print(record)
```

```python
def performance_category(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 80:
        return "VERY GOOD"
    elif score >= 70:
        return "GOOD"
    elif score >= 60:
        return "FAIR"
    else:
        return "POOR"

for record in records:
    record["Category"] = performance_category(record["Score"])

print("Performance Categories Added")
```

```python
print("ORGANIZED STUDENT/SUBJECT DATA\n")

for record in records:
    print(
        f"Student: {record['Student']} | "
        f"Categories: {record['Categories']} | "
        f"Score: {record['Score']} | "
        f"Category: {record['Category']}"
    )
```

```python
from collections import defaultdict

counter = defaultdict(int)

for record in records:
    key = (record["Student"], record["Categories"])
    counter[key] += 1

repeated_records = []

for key, count in counter.items():
    if count > 1:
        repeated_records.append(key)

print("REPEATED RECORDS\n")

for record in repeated_records:
    print(record)
```

```python
incomplete_records = []

for record in records:
    if record["Score"] is None:
        incomplete_records.append(record)

print("INCOMPLETE RECORDS\n")

for record in incomplete_records:
    print(record)
```

```python
student_summary = defaultdict(list)

for record in records:
    if record["Score"] is not None:
        student_summary[record["Student"]].append(record["Score"])

print("STUDENT SUMMARIES\n")

for student, scores in student_summary.items():

    average = sum(scores) / len(scores)

    print(f"\n{student}")
    print("Average:", round(average, 2))
    print("Highest:", max(scores))
    print("Lowest :", min(scores))
```

```python
subject_summary = defaultdict(list)

for record in records:
    if record["Score"] is not None:
        subject_summary[record["Categories"]].append(record["Score"])

print("CATEGORIES SUMMARIES\n")

for categories, scores in subject_summary.items():

    average = sum(scores) / len(scores)

    print(f"{categories}: Average Score = {average:.2f}")
```

```python
ordered_results = sorted(
    [r for r in records if r["Score"] is not None],
    key=lambda x: x["Score"],
    reverse=True
)

print("ORDERED RESULTS\n")

for record in ordered_results:
    print(
        f"{record['Student']} | "
        f"{record['Categories']} | "
        f"{record['Score']} | "
        f"{record['Category']}"
    )
```

```python
valid_scores = []

for record in records:
    if record["Score"] is not None:
        valid_scores.append(record["Score"])

overall_average = sum(valid_scores) / len(valid_scores)

print("=" * 50)
print("FINAL STUDENT PERFORMANCE REPORT")
print("=" * 50)

print("Total Records:", len(records))
print("Valid Records:", len(valid_scores))
print("Repeated Records:", len(repeated_records))
print("Incomplete Records:", len(incomplete_records))
print("Overall Average:", round(overall_average, 2))
```
